```python
import evaluate
# =========================
# BLEU Evaluation
# =========================

# Model-generated outputs (Predictions)
predictions = [
    "The project was completed successfully within the deadline",
    "The system encountered an unexpected error during execution"
]

# Reference sentences (multiple valid answers per prediction)
references = [
    [
        "The project was completed successfully within the deadline",
        "The project finished on time successfully"
    ],
    [
        "The system encountered an unexpected error during execution",
        "An unexpected error occurred while the system was running"
    ]
]

# Load BLEU metric
bleu = evaluate.load("bleu")

# Compute BLEU score
bleu_results = bleu.compute(predictions=predictions, references=references)

print("BLEU Results:")
print(bleu_results)
```

```python
# =========================
# ROUGE Evaluation
# =========================

# Model-generated summaries (Predictions)
predictions = [
    "The project was completed on time",
    "The system failed due to an unexpected error"
]

# Reference summaries (single reference per prediction)
references = [
    "The project was completed on time",
    "The system failed because of an unexpected error"
]

# Load ROUGE metric
rouge = evaluate.load("rouge")

# Compute ROUGE score
rouge_results = rouge.compute(predictions=predictions, references=references)

print("\nROUGE Results:")
print(rouge_results)
```
